```python
conn = sqlite3.connect("../sql/TTC_delay.db")

conn.execute("DROP TABLE IF EXISTS all_delays")

conn.execute("""
CREATE TABLE all_delays AS

SELECT
    'bus' AS transit_mode,
    date,
    time,
    day,
    month,
    hour,
    line,
    station,
    code,
    [Min Delay] AS min_delay,
    [Min Gap] AS min_gap,
    bound,
    vehicle
FROM bus_delays

UNION ALL

SELECT
    'streetcar' AS transit_mode,
    date,
    time,
    day,
    month,
    hour,
    line,
    station,
    code,
    [Min Delay] AS min_delay,
    [Min Gap] AS min_gap,
    bound,
    vehicle
FROM streetcar_delays

UNION ALL

SELECT
    'subway' AS transit_mode,
    date,
    time,
    day,
    month,
    hour,
    line,
    station,
    code,
    [Min Delay] AS min_delay,
    [Min Gap] AS min_gap,
    bound,
    vehicle
FROM subway_delays
""")

conn.commit()

mode_df = pd.read_sql("""
SELECT transit_mode, COUNT (*) AS rows
FROM all_delays
GROUP BY transit_mode
""", conn)

conn.close()
print(mode_df)
```

```python
# load lookup sheets
bus_lookup = pd.read_csv("../data/raw/ttc-bus-delay-data-lookup.csv", encoding="cp1252")

bus_lookup.head()
```

```python
stcar_lookup = pd.read_csv("../data/raw/ttc-streetcar-delay-data-lookup.csv", encoding="utf-8")

stcar_lookup.head()
```

```python
subway_lookup = pd.read_csv("../data/raw/ttc-subway-delay-data-lookup.csv", encoding="utf-8")

subway_lookup.head()
```

```python
for df in [bus_lookup, stcar_lookup, subway_lookup]:
    df.columns = df.columns.str.lower()

# add transit_mode to prevent clashing
bus_lookup["transit_mode"] = "bus"
stcar_lookup["transit_mode"] = "streetcar"
subway_lookup["transit_mode"] = "subway"

lookup_df = pd.concat([bus_lookup, stcar_lookup, subway_lookup], ignore_index=True)
lookup_df = lookup_df.drop_duplicates(subset=["transit_mode", "code"]).copy()

conn = sqlite3.connect("../sql/TTC_delay.db")
lookup_df.to_sql(
    name="delay_codes",
    con=conn,
    if_exists="replace",
    index=False
)

conn.close()
print("Combined lookup table successfully.")
print(f"{len(lookup_df)} unique rows")
```

```python
conn = sqlite3.connect("../sql/TTC_delay.db")

conn.execute("DROP VIEW IF EXISTS view_delays_by_mode")

conn.execute("""
CREATE VIEW view_delays_by_mode AS
SELECT
    transit_mode,
    COUNT (*) AS total_incidents,
    SUM(min_delay) AS total_delay_minutes,
    ROUND(AVG(min_delay), 2) AS avg_delay_minutes,
    MAX(min_delay) AS max_delay_minutes
FROM all_delays
GROUP BY transit_mode;
""")

conn.execute("DROP VIEW IF EXISTS view_delays_by_month")
conn.execute("""
CREATE VIEW view_delays_by_month AS
SELECT
    month,
    transit_mode,
    COUNT (*) AS total_incidents,
    SUM(min_delay) AS total_delay_minutes,
    ROUND(AVG(min_delay), 2) AS avg_delay_minutes
FROM all_delays
GROUP BY month, transit_mode;
""")

conn.execute("DROP VIEW IF EXISTS view_delays_by_hour")
conn.execute("""
CREATE VIEW view_delays_by_hour AS
SELECT
    hour,
    transit_mode,
    COUNT (*) AS total_incidents,
    SUM(min_delay) AS total_delay_minutes,
    ROUND(AVG(min_delay), 2) AS avg_delay_minutes
FROM all_delays
GROUP BY hour, transit_mode;
""")

conn.execute("DROP VIEW IF EXISTS view_top_lines")
conn.execute("""
CREATE VIEW view_top_lines AS
SELECT
    transit_mode,
    line,
    COUNT (*) AS total_incidents,
    SUM(min_delay) AS total_delay_minutes,
    ROUND(AVG(min_delay), 2) AS avg_delay_minutes
FROM all_delays
GROUP BY transit_mode, line
ORDER BY total_delay_minutes DESC;
""")

conn.execute("DROP VIEW IF EXISTS view_delay_reasons")
conn.execute("""
CREATE VIEW view_delay_reasons AS
SELECT
    a.transit_mode,
    COALESCE(b.description, 'Uncoded/Other') AS delay_reasons,
    COUNT (*) AS total_incidents,
    SUM(a.min_delay) AS total_delay_minutes,
    ROUND(AVG(a.min_delay), 2) AS avg_delay_minutes
FROM all_delays a
LEFT JOIN delay_codes b
    ON a.code = b.code AND a.transit_mode = b.transit_mode
GROUP BY a.transit_mode, delay_reasons
ORDER BY total_delay_minutes DESC;
""")

conn.commit()
conn.close()
print("Views updated successfully.")
```
